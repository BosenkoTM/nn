# Индивидуальное задание 2. Сверточные нейронные сети

## **Тема**. Классификация изображений (CNN).

## **Цель.** Обучить CNN для автоматического распознавания категорий одежды (компьютерное зрение для e-commerce).

## **Бизнес-задача**. Автоматическая категоризация товаров (одежды) для маркетплейса на основе датасета Fashion MNIST.



```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
import matplotlib.pyplot as plt
```

```python
# Загрузка датасета Fashion MNIST (товары: футболки, сумки, обувь и т.д.)
fashion_mnist = tf.keras.datasets.fashion_mnist
(train_images, train_labels), (test_images, test_labels) = fashion_mnist.load_data()

# Нормализация данных (масштабирование пикселей от 0 до 1)
train_images = train_images / 255.0
test_images = test_images / 255.0

# Изменение формы для сверточной сети (добавление канала цвета)
train_images = train_images.reshape(-1, 28, 28, 1)
test_images = test_images.reshape(-1, 28, 28, 1)
```

```python
# Построение архитектуры CNN
model_cnn = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(), # Переход от 2D к 1D вектору
    Dense(64, activation='relu'),
    Dense(10, activation='softmax') # 10 классов товаров
])

model_cnn.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
# Обучение
model_cnn.fit(train_images, train_labels, epochs=5, validation_data=(test_images, test_labels))
```

```python
# Визуализация предсказания для одного товара
import numpy as np
class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat', 'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

predictions = model_cnn.predict(test_images)
test_index = 0

plt.imshow(test_images[test_index].reshape(28, 28), cmap=plt.cm.binary)
plt.title(f"Предсказание сети: {class_names[np.argmax(predictions[test_index])]}")
plt.show()
```
