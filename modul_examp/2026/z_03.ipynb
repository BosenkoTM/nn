# Индивидуальное задание 3. Интеллектуальная обработка языка (NLP)


# **Цель.** Создать рекуррентную нейронную сеть (LSTM) для анализа тональности англоязычных отзывов.

## **Бизнес-задача**. Анализ отзывов клиентов (позитивный/негативный) для управления репутацией бренда.



```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

```python
# Загрузка датасета IMDB (отзывы на фильмы, но механика та же, что и для отзывов на товары)
vocab_size = 10000 # Ограничим словарь 10 000 самых частых слов
max_length = 150   # Длина отзыва (обрезаем или дополняем нулями)

(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=vocab_size)

# Приведение всех текстов к одной длине (padding)
X_train_pad = pad_sequences(X_train, maxlen=max_length, padding='post')
X_test_pad = pad_sequences(X_test, maxlen=max_length, padding='post')
```

```python
# Построение RNN (LSTM) модели
model_rnn = Sequential([
    Embedding(input_dim=vocab_size, output_dim=32, input_length=max_length), # Векторное представление слов
    LSTM(64, return_sequences=False), # Слой долгой краткосрочной памяти
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid') # 0 - негативный отзыв, 1 - позитивный
])

model_rnn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model_rnn.summary()
```

```python
# Обучение (в учебных целях берем малое количество эпох)
model_rnn.fit(X_train_pad, y_train, epochs=3, batch_size=64, validation_split=0.2)
```

```python
# Оценка качества модели
loss, accuracy = model_rnn.evaluate(X_test_pad, y_test)
print(f"Точность NLP модели: {accuracy*100:.2f}%")
```
